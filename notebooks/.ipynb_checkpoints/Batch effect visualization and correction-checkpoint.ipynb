```python
import umap.umap_ as umap
import os 
import pandas as pd
import seaborn as sns
import distinctipy
from combat.pycombat import pycombat

sns.set(style='white')

# Define the DATA_DIRECTORY and FIGURE_DIRECTORY
HERE = os. getcwd()
DATA_DIRECTORY = os.path.abspath(os.path.join(HERE, os.pardir, 'data'))
FIGURE_DIRECTORY = os.path.abspath(os.path.join(HERE, os.pardir, 'figures'))
```

```python
def UMAP_difford(df, phenotype_df, batch_name, title_name, dataset, bora):
    
    sns.set(style='white')
    # umap
    umap_reducer = umap.UMAP(n_components=2, init='random', random_state=0)
    umap_results = umap_reducer.fit_transform(df)
    # save the results
    visualization_umap = df.copy()
    visualization_umap['umap-one'] = umap_results[:,0]
    visualization_umap['umap-two'] = umap_results[:,1]
    # add batch metadata
    visualization_umap = pd.merge(visualization_umap, phenotype_df[[batch_name]], left_index=True, right_index=True, how='left')   
    
    g = sns.jointplot(data=visualization_umap, x="umap-one", y="umap-two", hue=batch_name, 
                 palette = sns.color_palette("tab20")[:len(set(visualization_umap[batch_name].values.tolist()))],
                )
    # g.ax_joint.legend(title=title_name,loc='upper right', borderaxespad=0)
    g.ax_joint.legend(title=title_name,loc='upper left', borderaxespad=0, bbox_to_anchor=(1.5, 0.1))
    g.savefig(os.path.join(FIGURE_DIRECTORY, dataset+'batch_correction_'+bora+'.png'))
    
    
    
def UMAP_plot_wallace_before(umap_results, df, cluster_name, dataset, bora):
    sns.set(style='white')
    visualization_umap = df.copy()
    visualization_umap['umap-one'] = umap_results[:,0]
    visualization_umap['umap-two'] = umap_results[:,1]
    # add batch metadata
    clusters=[]
    for i in visualization_umap['umap-two'].values.tolist():
        if i > 2 :
            clusters.append('2')
        else:
            clusters.append('1')
        
    g = sns.jointplot(data=visualization_umap, x="umap-one", y="umap-two", hue=clusters, 
                 palette = sns.color_palette('tab20')[9:11],
                )

    g.ax_joint.legend(title=cluster_name,loc='upper left', borderaxespad=0, bbox_to_anchor=(1.5, 0.1))
    g.savefig(os.path.join(FIGURE_DIRECTORY, dataset+'batch_correction_'+bora+'.png'))
 
    return clusters

def UMAP_plot_wallace_after(umap_results, df, clusters, title_name, dataset, bora):

    sns.set(style='white')
    visualization_umap = df.copy()
    visualization_umap['umap-one'] = umap_results[:,0]
    visualization_umap['umap-two'] = umap_results[:,1]
    # add batch metadata    

    g = sns.jointplot(data=visualization_umap, x="umap-one", y="umap-two", hue=clusters, 
                 palette = sns.color_palette('tab20')[9:11],
                )
    g.ax_joint.legend(title=title_name,loc='upper left', borderaxespad=0, bbox_to_anchor=(1.5, 0.1))
    g.savefig(os.path.join(FIGURE_DIRECTORY, dataset+'batch_correction_'+bora+'.png'))



def UMAP_plot_merged(df,batch_df, batch_name,title_name,dataset,bora):
    
    sns.set(style='white')
    # umap
    umap_reducer = umap.UMAP(n_components=2, init='random', random_state=0)
    umap_results = umap_reducer.fit_transform(df)
    # save the results
    visualization_umap = df.copy()
    visualization_umap['umap-one'] = umap_results[:,0]
    visualization_umap['umap-two'] = umap_results[:,1]
    # add batch metadata
    
    
    visualization_umap = pd.merge(visualization_umap, batch_df[[batch_name]], left_index=True, right_index=True, how='left')   
   
    hue_order=['Difford: sequencing batch 1', 'Difford: sequencing batch 2','Difford: sequencing batch 3',
              'Difford: sequencing batch 4', 'Difford: sequencing batch 5', 'Difford: sequencing batch 6',
              'Difford: sequencing batch 7','Difford: sequencing batch 8','Difford: sequencing batch 9',
              'Wallace: unknown cluster 2','Wallace: unknown cluster 1']

    
    g = sns.jointplot(data=visualization_umap, x="umap-one", y="umap-two", hue=batch_name, hue_order=hue_order,
                 palette = sns.color_palette("tab20")[:len(hue_order)]
                )

    g.ax_joint.legend(title=title_name,loc='upper left', borderaxespad=0, bbox_to_anchor=(1.5, 0.1))
    g.savefig(os.path.join(FIGURE_DIRECTORY, dataset+'batch_correction_'+bora+'.png'))
 


 
```

# 1. Batch effection visualization

## 1.1 Dataset Difford

```python
# loading the microbiome data
difford_X_filteredLow_relative_genus_standardized = pd.read_excel(os.path.join(DATA_DIRECTORY, 'interim','Difford2018', 'taxonomy', 'difford_X_filteredLow_relative_genus_standardized.xlsx'))
difford_X_filteredLow_relative_genus_standardized.set_index('metadata_sample_id', inplace=True)
difford_ids=difford_X_filteredLow_relative_genus_standardized.index.values.tolist()
difford_X_filteredLow_relative_genus_standardized
```

```python
# loading the corresponding metadata data
difford_metadata_path = os.path.join(DATA_DIRECTORY, 'raw','Difford2018', 'emission_metadata.tsv')
difford_metadata_df = pd.read_csv(difford_metadata_path, sep=' ', header=0)
difford_metadata_df= difford_metadata_df.sort_values(by=['sample_id'])

difford_metadata_df = difford_metadata_df.rename({"sample_id":"metadata_sample_id"}, axis=1)
difford_metadata_df = difford_metadata_df.set_index('metadata_sample_id')
difford_metadata_df = difford_metadata_df[difford_metadata_df.index.isin(difford_ids)]
difford_metadata_df
```

```python
UMAP_difford(difford_X_filteredLow_relative_genus_standardized, difford_metadata_df, 'Sequencing_Batch', 'Sequencing batch', 'difford', 'before')
```

```python
difford_batch_df = difford_metadata_df[['Sequencing_Batch']]
difford_batch_df.columns=[['batch']]
difford_batch_df['batch']='Difford: sequencing batch '+ difford_batch_df['batch'].astype(str)


difford_batch_df.to_csv(os.path.join(DATA_DIRECTORY, 'interim','Difford2018', 'difford_batch.tsv'), header=None)
difford_batch_df = pd.read_csv(os.path.join(DATA_DIRECTORY, 'interim','Difford2018', 'difford_batch.tsv'), index_col=[0], names=['batch'])
difford_batch_df
```

## 1.2 Dataset Wallace

```python
# loading the microbiome data
wallace_X_filteredLow_relative_genus_standardized = pd.read_excel(os.path.join(DATA_DIRECTORY, 'interim','Wallace2019', 'taxonomy', 'wallace_X_filteredLow_relative_genus_standardized.xlsx'))
wallace_X_filteredLow_relative_genus_standardized.set_index('metadata_sample_id', inplace=True)
wallace_ids=difford_X_filteredLow_relative_genus_standardized.index.values.tolist()
wallace_X_filteredLow_relative_genus_standardized
```

```python
umap_reducer = umap.UMAP(n_components=2, init='random', random_state=5)
umap_results = umap_reducer.fit_transform(wallace_X_filteredLow_relative_genus_standardized)
umap_results
```

```python
clusters= UMAP_plot_wallace_before(umap_results, wallace_X_filteredLow_relative_genus_standardized, 'Unknown cluster', 'wallace','before')
```

```python
wallace_batch_df = pd.DataFrame([["Wallace: unknown cluster "+ i] for i in clusters], index=wallace_X_filteredLow_relative_genus_standardized.index, columns =['batch'])

wallace_batch_df.to_csv(os.path.join(DATA_DIRECTORY, 'interim','Wallace2019', 'wallace_batch.tsv'), header=None)
wallace_batch_df = pd.read_csv(os.path.join(DATA_DIRECTORY, 'interim','Wallace2019', 'wallace_batch.tsv'), index_col=[0], names=['batch'])
wallace_batch_df
```
